```python
age_median = df["Age"].median()
df["Age"] = df["Age"].fillna(age_median)
age_median
```

```python
embarked_mode = df["Embarked"].mode()[0]
df["Embarked"] = df["Embarked"].fillna(embarked_mode)
embarked_mode
```

```python
df = df.drop("Cabin",axis=1)
df
```

```python
df.isnull().sum()
```

```python
df["Sex"] = df["Sex"].map({"male": 0, "female": 1})
df["Sex"]
```

```python
df.columns
```

```python
df = df.drop(columns=["Name", "Ticket"], errors="ignore")
df
```

```python
df.head()
```

```python
df.dtypes
```

```python
columns = ["Age", "Fare"]
for column in columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print("\nColumn:", column)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.boxplot(
    [df["Age"].dropna(), df["Fare"].dropna()],
    labels=["Age", "Fare"]
)

plt.title("Age and Fare - Before Outlier Removal")
plt.ylabel("Values")
plt.show()
```

```python
df_clean = df.copy()

for column in columns:
    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df_clean = df_clean[
        (df_clean[column] >= lower_bound) &
        (df_clean[column] <= upper_bound)
    ]

print("Original shape:", df.shape)
print("After outlier removal:", df_clean.shape)
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(
    [df_clean["Age"].dropna(), df_clean["Fare"].dropna()],
    labels=["Age", "Fare"]
)

plt.title("Age and Fare - After Outlier Removal")
plt.ylabel("Values")
plt.show()
```

```python
df_clean[columns] = (
    df_clean[columns] - df_clean[columns].mean()
) / df_clean[columns].std()

print("Standardization completed!")
```

```python
print("Mean:")
print(df_clean[columns].mean().round(2))

print("\nStandard Deviation:")
print(df_clean[columns].std().round(2))
```

```python
print("Final Dataset Shape:", df_clean.shape)

print("\nMissing Values:")
print(df_clean.isnull().sum())

print("\nData Types:")
print(df_clean.dtypes)

print("\nFirst 5 Rows:")
display(df_clean.head())
```
